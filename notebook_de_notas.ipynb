## 7. Week 3 - Class 2: Linear Regression from Scratch with Gradient Descent
*(linear_regression_from_scratch_tutorial.ipynb)*

Understanding the inner workings of linear modeling and optimization via gradient descent.

### 📌 Key Topics

* **Linear Prediction Equation:** Predictions are computed as:
  $$y_{pred} = w \cdot x + b$$
  where $w$ is the weight and $b$ is the bias.
* **Error and Cost (Mean Squared Error - MSE):**
  * **Loss:** The error for a single sample: $(y_{pred} - y)^2$.
  * **Cost:** The average of losses across the entire dataset:
    $$\text{MSE Cost} = \frac{1}{n} \sum_{i=1}^n (y_{pred} - y)^2$$
* **Gradient Descent:** Iterative optimization algorithm that updates $w$ and $b$ in the direction that decreases cost. Calculates partial derivatives:
  $$dw = \frac{2}{n} \sum X \cdot (y_{pred} - y)$$
  $$db = \frac{2}{n} \sum (y_{pred} - y)$$
* **Parameter Updates:**
  $$w \leftarrow w - \eta \cdot dw$$
  $$b \leftarrow b - \eta \cdot db$$
  where $\eta$ is the **learning rate**.
* **Learning Rate ($\eta$) Behavior:**
  * *Too small:* Training progresses extremely slowly.
  * *Just right:* Cost steadily decreases each epoch until convergence.
  * *Too large:* The algorithm overshoots, diverges, and cost explodes (values become huge or `NaN`).

Using the industry standard library to train, evaluate, and inspect robust linear models.

### 📌 Key Topics

* **Standard scikit-learn Workflow:**
  1. Separate features ($X$) and target ($y$).
  2. Split with `train_test_split` using a fixed `random_state`.
  3. Instantiate the model: `LinearRegression()`.
  4. Fit to the training data using `.fit(X_train, y_train)`.
  5. Make predictions using `.predict(X_test)`.
* **Evaluation Metrics:**
  * **MAE (Mean Absolute Error):** Average absolute error (interpretable in original units).
  * **RMSE (Root Mean Squared Error):** Square root of the mean squared error (penalizes larger errors more heavily).
  * **$R^2$ (R-squared):** Proportion of variance in target explained by the model (ranges from $-\infty$ to $1.0$).
* **Model Inspection:**
  * `model.coef_`: Weight associated with each input feature.
  * `model.intercept_`: Bias term ($b$).

### ⚠️ Common Pitfalls to Avoid
1. **Fitting on the test set:** Only call `.fit()` on training data. Doing it on the test set inflates performance metrics falsely.
2. **Forgetting `random_state`:** Prevents your train/test splits from being reproducible.
3. **Data leakage before splitting:** Computing dataset-wide statistics (like mean/std) before splitting. Splitting must happen first.
4. **Reporting only training metrics:** A model can score perfectly on training data but perform poorly on unseen data (*overfitting*).

## 12. Week 5 - Class 1: Logistic Regression from Scratch
*(logistic_regression_from_scratch.ipynb)*

### 📌 Key Topics

* **Linear Combination:**
  $$z = X \cdot w + b$$
  In code: `z = X @ self.weights + self.bias`

* **Sigmoid Function:**
  Maps raw predictions ($z$) to probabilities between 0 and 1:
  $$\sigma(z) = \frac{1}{1 + e^{-z}}$$
  In code: `1 / (1 + np.exp(-z))`

* **Log Loss (Binary Cross-Entropy):**
  Classification cost function — measures how wrong the **probabilities** are (not just 0/1 accuracy):
  $$\text{Log Loss} = -\frac{1}{n} \sum_{i=1}^{n} \left[ \, y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \, \right]$$
  * Heavily penalizes being **confidently wrong** (e.g. predicting $0.01$ when the true label is $1$).
  * Lower log loss = better calibrated probabilities.

* **Decision Boundary:**
  Geometric boundary separating the classes, occurring at prediction probability = $0.5$ ($z=0$).

* **Zero Initialization:**
  `self.weights = np.zeros(n_features)`. Optimal for linear and logistic regression because their loss functions are convex, guaranteeing convergence to the global minimum.

### ❓ Doubts Clarified

* **What is log loss, simply?** It scores how good the model's *chances* are. If the customer churned and the model said 90%, loss is low; if it said 10%, loss is high. Being wrong with 99% confidence hurts much more than being wrong with 51%.

### 💻 Reference Code


```python
import numpy as np

class LogisticRegressionFromScratch:
    def __init__(self, lr=0.01, n_iters=1000):
        self.lr = lr
        self.n_iters = n_iters
        self.weights = None
        self.bias = 0.0

    def _sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0

        for _ in range(self.n_iters):
            z = X @ self.weights + self.bias
            y_pred = self._sigmoid(z)

            dw = (1 / n_samples) * (X.T @ (y_pred - y))
            db = (1 / n_samples) * np.sum(y_pred - y)

            self.weights -= self.lr * dw
            self.bias -= self.lr * db

    def predict_proba(self, X):
        z = X @ self.weights + self.bias
        return self._sigmoid(z)

    def predict(self, X):
        probabilities = self.predict_proba(X)
        return np.where(probabilities >= 0.5, 1, 0)
```

### 📌 Key Topics

* **Class imbalance & accuracy baseline:** ~26.5% of customers churned → ~73.5% did not. A dumb model that always predicts "not churned" already gets **73.5% accuracy** without learning anything. Do not trust accuracy alone when classes are uneven.

* **Stratified train/test split (`stratify=y`):** Applies the 80/20 split *inside each class*, so train and test keep the same churn rate (~26.6%). Without it, a random split can put more churners in one set than the other — unfair evaluation. Matters more when the dataset is small or the positive class is rare.

* **Feature scaling (`StandardScaler`):** Logistic regression's gradient solver is sensitive to scale. Features live on very different ranges (e.g. `SeniorCitizen` is 0/1, `tenure` is 1–72 months, `TotalCharges` reaches thousands). Without scaling, large-magnitude features dominate the gradient. After scaling, a "+1" on a feature means **+1 standard deviation**, not +1 raw unit.

* **`predict()` vs `predict_proba()`:**
  * `predict_proba(X)` → raw probabilities per class; second column is P(churn).
  * `predict(X)` → hard 0/1 after applying the **0.5** threshold.
  * Same model; one answers "how sure?", the other "final label".

* **Interpreting `model.coef_`:** One weight per feature (same order as columns). Positive coef → feature up → log-odds of churn up → probability of churn up. Negative → away from churn. Log-odds = the internal score $z$ before the sigmoid.

* **Regularization & the `C` parameter (L2 by default):**
  Minimizes: $\sum \text{log loss} + \frac{1}{2C}\sum w^2$
  * **Small C** (e.g. 0.01): strong regularization → weights pushed toward zero → less overfitting, risk of underfitting.
  * **Large C** (e.g. 100): weak regularization → large weights allowed → fit training data more closely.
  * `C` feels backwards because it sits in the **denominator** of the penalty.
  * Another reason to scale first: unscaled small-unit features need large weights and get unfairly punished.
  * Classic overfitting signal: **wide gap** between train accuracy and test accuracy.

### ❓ Doubts Clarified

* **What did the 26.5% note mean?** Classes are imbalanced. Always-predict-majority already scores ~73.5%. A model at 75% accuracy may barely beat that baseline.

* **What did the stratified-split note mean?** `test_size=0.2` with `stratify=y` keeps 80% of *churners* and 80% of *non-churners* in train. That is why train/test/full churn rates all match (~0.266).

* **`tenure` vs `SeniorCitizen`?** Different meanings *and* scales: `SeniorCitizen` = elderly flag (0/1); `tenure` = months as customer (1–72). The notebook highlights them to show why scaling is needed.

* **`predict` vs `predict_proba`?** Probabilities first (e.g. 72% churn); `predict` just cuts at 0.5 → returns 1. If prob is 0.4 → predict returns 0.

* **Coefficients in plain words?** Each feature gets a weight. Positive weight pushes toward churn; negative pushes toward stay. After scaling, interpret "+1" as one std above the mean.

* **What is regularization / C?** A penalty on large weights so the model does not memorize noise. Small C = strong brake; large C = loose brake.

* **What is log loss?** How wrong the *probabilities* are. Confident mistakes cost the most.

Accuracy alone is misleading when classes are imbalanced. The **confusion matrix** (TP, TN, FP, FN) is the foundation for precision, recall, F1, and threshold tuning.

### 📌 Key Topics

* **Confusion Matrix:**
  |  | Predicted Negative | Predicted Positive |
  |--|--|--|
  | **Actual Negative** | TN (correct stay) | FP (false alarm) |
  | **Actual Positive** | FN (missed) | TP (caught) |

* **Accuracy** = (TP + TN) / total — fraction of all predictions correct. Good on balanced data; misleading when one class dominates.

* **Precision** = TP / (TP + FP) — of all flagged positives, how many were truly positive? Important when **false positives are costly** (e.g. expensive retention offer to someone who wasn't leaving).

* **Recall** = TP / (TP + FN) — of all actual positives, how many did we catch? Important when **false negatives are costly** (e.g. missing a churner / fraud case).

* **F1 score** = 2 × (Precision × Recall) / (Precision + Recall) — harmonic mean; one number when you care about both.

* **Higher is better:** Accuracy, Precision, Recall, F1. **Lower is better:** FPR, FNR, error rate.

* **Flagged:** how many customers the model marked as positive (churn) at that threshold — `yp.sum()`.

* **Threshold trade-off:** model outputs P(churn) ∈ [0, 1]. Predict positive if P ≥ threshold.
  * **Lower threshold** → more flagged → higher recall, lower precision
  * **Higher threshold** → fewer flagged → higher precision, lower recall
  * No universal best threshold — depends on FP vs FN cost. To maximize F1 on this churn model, ~0.21 beat the default 0.5.

### ❓ Doubts Clarified

* **How do I pick the best threshold?** If maximizing F1 → pick the threshold with highest F1. If catching positives matters more (fraud/churn) → lower it. If false alarms are costly → raise it.

* **What does the Precision/Recall vs Threshold graph show?** Blue = precision rises as threshold rises; coral = recall falls. Trade-off visualized; 0.5 is only the default cut, not always optimal.
