### 📌 Key Topics

* **Correlation Matrix:** Checking linear dependencies using `df.corr(numeric_only=True)`. Correlation values range from $-1$ to $+1$, where values near $0$ signify no linear relationship.
* **Skewness:** Evaluating asymmetry of distributions using `.skew()`. Highly right-skewed variables (skewness $> 1.0$) typically benefit from a log transform prior to modeling.
* **Outlier Detection via IQR:** Detecting outliers mathematically using the Interquartile Range:
  $$\text{IQR} = Q_3 - Q_1$$
  $$\text{Lower Bound} = Q_1 - 1.5 \times \text{IQR}$$
  $$\text{Upper Bound} = Q_3 + 1.5 \times \text{IQR}$$

```python
# Check target skewness
salary_skew = df['salary_in_usd'].skew()
print(f"Skewness of Target: {salary_skew:.3f}")

# IQR outlier limits
q1 = df['salary_in_usd'].quantile(0.25)
q3 = df['salary_in_usd'].quantile(0.75)
iqr = q3 - q1
upper_bound = q3 + 1.5 * iqr
lower_bound = q1 - 1.5 * iqr
```

## 7. Week 3 - Class 2: Linear Regression from Scratch with Gradient Descent
*(linear_regression_from_scratch_tutorial.ipynb)*

Understanding the inner workings of linear modeling and optimization via gradient descent.

### 📌 Key Topics

* **Linear Prediction Equation:** Predictions are computed as:
  $$y_{pred} = w \cdot x + b$$
  where $w$ is the weight and $b$ is the bias.
* **Error and Cost (Mean Squared Error - MSE):**
  * **Loss:** The error for a single sample: $(y_{pred} - y)^2$.
  * **Cost:** The average of losses across the entire dataset:
    $$\text{MSE Cost} = \frac{1}{n} \sum_{i=1}^n (y_{pred} - y)^2$$
* **Gradient Descent:** Iterative optimization algorithm that updates $w$ and $b$ in the direction that decreases cost. Calculates partial derivatives:
  $$dw = \frac{2}{n} \sum X \cdot (y_{pred} - y)$$
  $$db = \frac{2}{n} \sum (y_{pred} - y)$$
* **Parameter Updates:**
  $$w \leftarrow w - \eta \cdot dw$$
  $$b \leftarrow b - \eta \cdot db$$
  where $\eta$ is the **learning rate**.
* **Learning Rate ($\eta$) Behavior:**
  * *Too small:* Training progresses extremely slowly.
  * *Just right:* Cost steadily decreases each epoch until convergence.
  * *Too large:* The algorithm overshoots, diverges, and cost explodes (values become huge or `NaN`).

*(linear_regression_sklearn_tutorial.ipynb)*

Using the industry standard library to train, evaluate, and inspect robust linear models.

### 📌 Key Topics

* **Standard scikit-learn Workflow:**
  1. Separate features ($X$) and target ($y$).
  2. Split with `train_test_split` using a fixed `random_state`.
  3. Instantiate the model: `LinearRegression()`.
  4. Fit to the training data using `.fit(X_train, y_train)`.
  5. Make predictions using `.predict(X_test)`.
* **Evaluation Metrics:**
  * **MAE (Mean Absolute Error):** Average absolute error (interpretable in original units).
  * **RMSE (Root Mean Squared Error):** Square root of the mean squared error (penalizes larger errors more heavily).
  * **$R^2$ (R-squared):** Proportion of variance in target explained by the model (ranges from $-\infty$ to $1.0$).
* **Model Inspection:**
  * `model.coef_`: Weight associated with each input feature.
  * `model.intercept_`: Bias term ($b$).

### ⚠️ Common Pitfalls to Avoid
1. **Fitting on the test set:** Only call `.fit()` on training data. Doing it on the test set inflates performance metrics falsely.
2. **Forgetting `random_state`:** Prevents your train/test splits from being reproducible.
3. **Data leakage before splitting:** Computing dataset-wide statistics (like mean/std) before splitting. Splitting must happen first.
4. **Reporting only training metrics:** A model can score perfectly on training data but perform poorly on unseen data (*overfitting*).

## 12. Week 5: Logistic Regression from Scratch
*(logistic_regression_from_scratch.ipynb)*

### 📌 Key Topics

* **Linear Combination:**
  $$z = X \cdot w + b$$
  In code: `z = X @ self.weights + self.bias`

* **Sigmoid Function:**
  Maps raw predictions ($z$) to probabilities between 0 and 1:
  $$\sigma(z) = \frac{1}{1 + e^{-z}}$$
  In code: `1 / (1 + np.exp(-z))`

* **Log Loss (Binary Cross-Entropy):**
  Evaluation cost function:
  $$\text{Log Loss} = -\frac{1}{n} \sum_{i=1}^{n} \left[ \, y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \ y_i) \, \right]$$
  * Heavily penalizes the model for being confidently wrong (e.g. predicting $0.01$ probability for a true label of $1$).

* **Decision Boundary:**
  Geometric boundary separating the classes, occurring at prediction probability = $0.5$ ($z=0$).

* **Zero Initialization:**
  `self.weights = np.zeros(n_features)`. Optimal for linear and logistic regression because their loss functions are convex, guaranteeing convergence to the global minimum.

### 💻 Reference Code

```python
import numpy as np

class LogisticRegressionFromScratch:
    def __init__(self, lr=0.01, n_iters=1000):
        self.lr = lr
        self.n_iters = n_iters
        self.weights = None
        self.bias = 0.0

    def _sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0.0

        for _ in range(self.n_iters):
            z = X @ self.weights + self.bias
            y_pred = self._sigmoid(z)

            dw = (1 / n_samples) * (X.T @ (y_pred - y))
            db = (1 / n_samples) * np.sum(y_pred - y)

            self.weights -= self.lr * dw
            self.bias -= self.lr * db

    def predict_proba(self, X):
        z = X @ self.weights + self.bias
        return self._sigmoid(z)

    def predict(self, X):
        probabilities = self.predict_proba(X)
        return np.where(probabilities >= 0.5, 1, 0)
```
